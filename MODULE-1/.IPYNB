## A. What Is a Neural Network?

A neural network is made of connected computational units called **neurons**.

Each neuron usually does three things:

1. Receives input values
2. Multiplies them by adjustable numbers called **weights**
3. Applies an activation function to produce an output

A simplified neuron pipeline:
**Inputs -> Weighted sum -> Activation function -> Output**

Mathematically:
- $z = w_1x_1 + w_2x_2 + ... + w_nx_n + b$
- $a = g(z)$

Where:
- $x$: input features
- $w$: weights
- $b$: bias
- $z$: weighted sum
- $g(z)$: activation function
- $a$: output of the neuron

## Neural Network Layers

Neural networks are organized into layers:
**Input layer -> Hidden layer(s) -> Output layer**

Why do layers matter?
For image recognition, layers often learn features in stages:

- Early layers: edges
- Middle layers: shapes or textures
- Later layers: objects (faces, cars, cats, etc.)

This layered feature learning is a major reason neural networks are powerful.

## Key Terms

| Term                | Meaning                                                   |
| ------------------- | --------------------------------------------------------- |
| Feature             | An input variable used by the model                       |
| Weight              | A learned number showing how important an input is        |
| Bias                | An extra learned value that shifts the neuron’s output    |
| Activation function | A function that adds nonlinearity                         |
| Hidden layer        | A layer between input and output                          |
| Parameter           | A value learned during training, such as a weight or bias |


## Why Are Weights Needed?

The role of a **weight** is to control how much each input feature affects the model’s prediction.

For an input $x_i$, its contribution is:

$$
w_i x_i
$$

So the weight $w_i$ does three things:

* **Strength:** larger absolute value means a stronger influence.
* **Direction:** positive pushes the prediction upward; negative pushes it downward.
* **Learning:** during training, the model adjusts weights to reduce prediction errors.

Example:

$$
z = 2(\text{study hours}) - 3(\text{missed classes}) + b
$$

Here, study hours have a positive effect, while missed classes have a negative effect. The weights (2) and (-3) determine those effects.


## Why Is an Activation Function Needed?

Neural networks need **activation functions** mainly to add **nonlinearity**.

Common examples:

* **ReLU:** $f(z) = \max(0, z)$ - widely used in hidden layers.
* **Sigmoid:** outputs values between 0 and 1 — often used for binary-classification output.
* **Softmax:** converts several output scores into class probabilities — used for multi-class classification.
* **Tanh:** outputs between -1 and 1.

So the main role is:

> Activation functions let neural networks learn complex, non-linear relationships from data.

```python

#* a single artificial neuron

import numpy as np
#? House price prediction
x = np.array([1200, 4])
w = np.array([.2,.4])
b = .6
z = np.dot(w,x) + b
print("Neuron output :", z)
```

```python

#* Activation function

import numpy as np 

def sigmoid(z):
    return 1 / (1 + np.exp(-z))
def relu(z):
    return np.maximum(0,z)

values = np.array([-1,-2,-3,1,2,3,0])

print("Sigmoid", sigmoid(values))
print("relu", relu(values))
```

```python

#* Simple binary classification prediction(Wether a student pass or fail)

import numpy as np 
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

hour_studied = 5
weight = 2
bias = 1

z = weight * hour_studied + bias 
prob_of_passing  = sigmoid(z)
print("prob of passing : ", prob_of_passing)

if prob_of_passing >= .5:
    print("Pass")
else:
    print("Fail")
```

```python

#* A tiny neural network with one hidden layer

import numpy as np 

def relu(z):
    return np.maximum(0,z)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

x = np.array([1,2])

#? Input layer 
w1 = np.array([[0.2,0.3],
               [-0.4,0.5]])
b1 = np.array([0.3,-0.4])

#? Hidden layer 
w2 = np.array([0.4,0.5])
b2 = 0.4

z1 = np.dot(w1,x) + b1
A1 = relu(z1)

z2 = np.dot(w2,A1) + b2
A2 = sigmoid(z2)

print("Hidden layer values ",A1)
print("Output layer values ",A2)
```
